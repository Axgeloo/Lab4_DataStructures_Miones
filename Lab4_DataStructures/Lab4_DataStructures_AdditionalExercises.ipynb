```python
LAST_NAME = "Miones"
SEED_NUM = 2
FAVORITE_ARTIST = "Cup of Joe"

sources = ["T1", "T2", "T3", "T4", "T5"]
measurements = ["TEMPERATURE", "VOLTAGE", "CURRENT",
                "PRESSURE", "SPEED"]

artist = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]
if not artist:
    artist = [65]

records = []

# Generate test measurements
for i, source in enumerate(sources):
    for j, measurement in enumerate(measurements):
        value = 1 + (
            SEED_NUM + artist[(i + j) % len(artist)]
            + i * 13 + j * 7
        ) % 100

        records.append({
            "Source": source,
            "Measurement": measurement,
            "Value": value
        })

# Add duplicate
records.append(records[0].copy())

# Detect duplicates
seen = set()
duplicates = []

for record in records:
    key = (record["Source"], record["Measurement"])
    if key in seen:
        duplicates.append(record)
    else:
        seen.add(key)

print("TEST RECORDS")
for record in records:
    print(record)

print("\nDUPLICATES")
for record in duplicates:
    print(record)

# Summarize by measurement category
print("\nMEASUREMENT SUMMARY")
for measurement in measurements:
    values = [
        r["Value"] for r in records
        if r["Measurement"] == measurement
        and (r["Source"], r["Measurement"]) not in {
            (d["Source"], d["Measurement"]) for d in duplicates
        }
    ]

    print(measurement,
          "Count:", len(values),
          "Average:", round(sum(values) / len(values), 2),
          "Minimum:", min(values),
          "Maximum:", max(values))

# Sort records
print("\nSORTED RECORDS")
for record in sorted(records, key=lambda r: (r["Source"], r["Measurement"])):
    print(record)

print("\nFINAL ENGINEERING REPORT")
print("Total records:", len(records))
print("Duplicate records:", len(duplicates))
print("Test sources:", len(sources))
print("Measurement categories:", len(measurements))
```

```python
from collections import Counter

LAST_NAME = "Miones"
SEED_NUM = 2
FAVORITE_ARTIST = "Cup of Joe"

students = [f"{LAST_NAME}_{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS",
            "PHYSICS", "ENGLISH"]

records = []
artist = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]
if not artist:
    artist = [65]

# Generate student scores
for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        score = 60 + (
            SEED_NUM + artist[(i + j) % len(artist)]
            + i * 7 + j * 11
        ) % 41

        records.append({
            "Student": student,
            "Subject": subject,
            "Score": score
        })

# Add a duplicate and an incomplete record
records.append(records[0].copy())
records.append({
    "Student": students[0],
    "Subject": "MATHEMATICS",
    "Score": None
})

# Detect duplicates
seen = set()
duplicates = []

for record in records:
    key = (record["Student"], record["Subject"])
    if key in seen:
        duplicates.append(record)
    else:
        seen.add(key)

# Categorize scores
def category(score):
    if score is None:
        return "Incomplete"
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 75:
        return "Satisfactory"
    return "Needs Improvement"

for record in records:
    record["Category"] = category(record["Score"])

# Display records
print("STUDENT RECORDS")
for record in records:
    print(record)

print("\nDUPLICATE RECORDS")
for record in duplicates:
    print(record)

print("\nINCOMPLETE RECORDS")
for record in records:
    if record["Score"] is None:
        print(record)

# Summarize student performance
print("\nSTUDENT SUMMARY")
for student in students:
    scores = [
        r["Score"] for r in records
        if r["Student"] == student and r["Score"] is not None
        and (r["Student"], r["Subject"]) not in {
            (d["Student"], d["Subject"]) for d in duplicates
        }
    ]

    if scores:
        print(student,
              "Average:", round(sum(scores) / len(scores), 2),
              "Highest:", max(scores),
              "Lowest:", min(scores))

# Category summary
counts = Counter(
    r["Category"] for r in records
    if r["Score"] is not None
    and (r["Student"], r["Subject"]) not in {
        (d["Student"], d["Subject"]) for d in duplicates
    }
)

print("\nCATEGORY SUMMARY")
for name, count in counts.items():
    print(name, ":", count)

# Final report
valid_scores = [
    r["Score"] for r in records
    if r["Score"] is not None
    and (r["Student"], r["Subject"]) not in {
        (d["Student"], d["Subject"]) for d in duplicates
    }
]

print("\nFINAL REPORT")
print("Total records:", len(records))
print("Duplicates:", len(duplicates))
print("Incomplete records:", sum(r["Score"] is None for r in records))
print("Overall average:", round(sum(valid_scores) / len(valid_scores), 2))
```

```python
LAST_NAME = "Miones"
SEED_NUM = 2
FAVORITE_ARTIST = "Cup of Joe"

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

prices = {
    "NOTEBOOK": 35.00,
    "PEN": 15.00,
    "USB": 250.00,
    "HEADPHONES": 450.00,
    "CALCULATOR": 180.00
}

products = list(prices.keys())
artist = [ord(c) for c in FAVORITE_ARTIST if c.isalpha()]
if not artist:
    artist = [65]

transactions = []

# Generate transactions
for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        quantity = 1 + (
            SEED_NUM + artist[(i + j) % len(artist)] + i + j
        ) % 5

        transactions.append({
            "Customer": customer,
            "Product": product,
            "Quantity": quantity,
            "Unit Price": prices[product],
            "Total": quantity * prices[product]
        })

# Add duplicate transaction
transactions.append(transactions[0].copy())

# Detect repeated customer-product records
seen = set()
duplicates = []

for transaction in transactions:
    key = (transaction["Customer"], transaction["Product"])
    if key in seen:
        duplicates.append(transaction)
    else:
        seen.add(key)

print("TRANSACTION RECORDS")
for transaction in transactions:
    print(transaction)

print("\nDUPLICATE PURCHASES")
for transaction in duplicates:
    print(transaction)

# Summarize spending per customer
print("\nCUSTOMER SUMMARY")
for customer in customers:
    customer_records = [
        t for t in transactions
        if t["Customer"] == customer
        and (t["Customer"], t["Product"]) not in {
            (d["Customer"], d["Product"]) for d in duplicates
        }
    ]

    total = sum(t["Total"] for t in customer_records)
    quantity = sum(t["Quantity"] for t in customer_records)

    print(customer, "Spending:", total, "Units:", quantity)

# Summarize sales per product
print("\nPRODUCT SUMMARY")
for product in products:
    product_records = [
        t for t in transactions
        if t["Product"] == product
        and (t["Customer"], t["Product"]) not in {
            (d["Customer"], d["Product"]) for d in duplicates
        }
    ]

    quantity = sum(t["Quantity"] for t in product_records)
    sales = sum(t["Total"] for t in product_records)

    print(product, "Quantity sold:", quantity, "Total sales:", sales)

# Final report
valid = [
    t for t in transactions
    if (t["Customer"], t["Product"]) not in {
        (d["Customer"], d["Product"]) for d in duplicates
    }
]

print("\nFINAL TRANSACTION REPORT")
print("Total records:", len(transactions))
print("Duplicate records:", len(duplicates))
print("Distinct customers:", len(set(t["Customer"] for t in transactions)))
print("Distinct products:", len(set(t["Product"] for t in transactions)))
print("Total sales:", sum(t["Total"] for t in valid))
```
